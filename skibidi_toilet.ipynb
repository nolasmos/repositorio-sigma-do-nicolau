criar um programa que identifica se o numero é par ou nao

```python
#entrada
numero = int(input("digite um numero: "))
#processamento
if numero % 2 == 0:
    print(f"o numero {numero} é par")
else:
    print(f"o numero {numero} é impar")
```

escrever um programa que recebe 4 notas referentes aos bimestras, calcula a media e mostra se ele foi aprovdo, ficou de recuperaçao ou foi reprovado

```python
nota1 = float(input("digite a primeira nota: "))
nota2 = float(input("digite a segunda nota: "))
nota3 = float(input("digite a terceira nota: "))
nota4 = float(input("digite a quarta nota: "))
nome = input("digite o nome do aluno: ")
media = (nota1 + nota2 + nota3 + nota4) / 4
if nota1 < 0 or nota2 < 0 or nota3 < 0 or nota4 < 0:
    print("nota invalida")
if nota1 > 10 or nota2 > 10 or nota3 > 10 or nota4 > 10:
    print("nota invalida")
elif media >= 7:
    print(f"parabens {nome}, voce foi aprovado")
elif media >= 5 and media < 7:
    print(f"{nome}, voce ficou de recuperaçao")
elif media <5:
  print(f"{nome}, voce foi reprovado")
```
