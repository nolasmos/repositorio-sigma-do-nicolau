```python
#escrever um programa que recebe 4 notas do aluno, calcula a média e mostra se foi aprovado

#entrada
n1 = float(input("Digite a primeira nota: "))
n2 = float(input("Digite a segunda nota: "))
n3 = float(input("Digite a terceira nota: "))
n4 = float(input("Digite a quarta nota: "))

#processamento
media = (n1 + n2 + n3 + n4) / 4

if media >= 7:
  print (f'foi aprovado com a media de {media}')
elif media >= 5 and media < 7:
  print(f"você ficou de recuperação com a media de {media}")
else:
  print(f'você foi reprovado com a media de {media}')

```
