```python
numero = 2
match numero:
  case 1:
    print("você escolheu o número 1")
  case 2:
    print("você escolheu o número 2")
  case 3:
    print("você escolheu o número 3")
```

```python
cor = "azul"

match cor:
  case "azul":
    print("Cor azul")
  case _:
    print("Outra cor")
```

```python
numero = 2

match numero:
  case 1:
    print("Um")
  case 2:
    print("Dois")
  case 3:
    print("Três")
  case _:
    print("Número não mapeado")
```

```python
opcao = 2

match opcao:
  case 1:
    print("Água")
  case 2:
    print("Suco")
  case 3:
    print("Refrigerante")
  case _:
    print("Opção inválida")
```

```python
dia = "sábado"

match dia:
  case "sábado" | "domingo":
    print("Final de semana")
  case (
      "segunda" | "terça" | "quarta" | "quinta" | "sexta"
  ):
    print("Dia de semana")
  case _:
    print("Outro dia / Dia inválido")
```

```python
nota = "A"

match nota:
  case "A":
    print("Excelente")
  case "B":
    print("Bom")
  case "C":
    print("Precisa melhorar")
  case _:
    print("Nota inválida")
```
