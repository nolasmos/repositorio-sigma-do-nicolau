```python
def caixa_eletronico():
    try:

        valor = int(input("Digite o valor do saque: "))

        if valor < 0:
            print("Por favor, digite um valor maior que zero.")
            return

        notas = [100, 50, 20, 10, 5, 2]
        resultado = {}
        valor_restante = valor


        for nota in notas:
            quantidade = valor_restante // nota
            if quantidade > 0:
                resultado[nota] = quantidade
                valor_restante %= nota



        else:
            print("Saída esperada:")
            for nota in notas:
                if nota in resultado:
                    print(f"{resultado[nota]} nota(s) de R${nota}")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")


caixa_eletronico()
```
