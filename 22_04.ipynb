```python
#escrever um programa que o usuario digite uma palavra e verifique se é palindroma
palavra = input("digite uma palavra: ")
invertida = ""

for i in (palavra):
 invertida = i + invertida
print(invertida)

if palavra == invertida:
  print("é uma palavra palindroma")
else:
  print("nao é uma palavra palindroma")
```
