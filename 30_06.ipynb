```python
#cofre digital
#o codigo possui 5 numeros
#leia os 5 numeros com laço de repetiçao
#quantos numeros sao pares/impares
#o maior/menor numero
#a soma e media dos numeros

pares = 0
impares = 0
soma = 0

numero = int(input("Digite o 1º número: "))
maior = numero
menor = numero
soma += numero

if numero % 2 == 0:
    pares += 1
else:
    impares += 1

for i in range(2, 6):
    numero = int(input(f"Digite o {i}º número: "))
    soma += numero

    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1

    if numero > maior:
        maior = numero

    if numero < menor:
        menor = numero

media = soma / 5

print("\n=== Resultado ===")
print("Quantidade de pares:", pares)
print("Quantidade de ímpares:", impares)
print("Maior número:", maior)
print("Menor número:", menor)
print("Soma dos números:", soma)
print("Média dos números:", media)
```

```python
#numero espelho
#peça um numero, inverta e veja se ele é palindromo
numero = int(input("Digite um número: "))
numero_invertido = int(str(numero)[::-1])
print("Número invertido:", numero_invertido)

if numero == numero_invertido:
    print("O número é um palíndromo.")
else:
    print("O número não é um palíndromo")
```
