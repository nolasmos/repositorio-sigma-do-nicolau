```python
#você foi contratado para desenvolver um sistema simles de caixa eletronico
#o sistema deverá funcionar através de um menu interativo utilizando a estrutura de repetiçao while
#o programa devera deve iniciar com um saldo inicial de 1000 e permitir que o usuario escolha diferentes operações até decidir sair do sistema
#funcionalidades
#1 ver saldo
#2 depositar dinheiro
#3 sacar dimheiro
#4 sair
#regras
#ver saldo
saldo = 1000.00
opcao = 0

while opcao != 4:
    print("1 - Ver saldo")
    print("2 - Depositar dinheiro")
    print("3 - Sacar dinheiro")
    print("4 - Sair")

    opcao = int(input("Digite a opção desejada: "))

    if opcao == 1:
      print(f"Seu saldo é igual a {saldo}")

    if opcao == 2:
      valor_deposito = float(input("Digite o valor do seu deposito: "))
      saldo = saldo + valor_deposito
      print(f"")
      print(f"")
      print(f"Você depositou R${valor_deposito}")
      print(f"Seu saldo agora é de: R${saldo}")
      print(f"")
      print(f"")

    if opcao == 3:
      valor_sacar = float(input("Digite o valor que quer sacar: "))
      saldo = saldo - valor_sacar
      print(f"")
      print(f"")
      print(f"Você sacou R${valor_sacar}")
      print(f"Seu saldo agora é de R${saldo}")
      print(f"")
      print(f"")

    if opcao == 4:
      print(f"")
      print(f"OBRIGADO POR USAR ESTE CAIXA ELETRÔNICO!")
```
